```python
import numpy as np
import pandas as pd
```

```python
from tensorflow.keras.datasets import cifar10
(X_train,y_train),(X_test,y_test) = cifar10.load_data()
```

```python
X_train_rows = X_train.reshape(X_train.shape[0],-1)
X_test_rows = X_test.reshape(X_test.shape[0],-1)
```

```python
class NearestNeighbour:
    def __init__(self):
        pass
    
    def train(self,X,y):
        self.Xtr = X
        self.ytr = y.flatten()
    
    def predict(self,X):
        ex = X.shape[0]
        y_pred = np.zeros(ex)

        
        for i in range(ex):
            distances = np.sum(np.abs(self.Xtr - X[i,:]),axis=1)
            min_idx = np.argmin(distances)
            y_pred[i]=self.ytr[min_idx]

        return y_pred
```

```python
nn = NearestNeighbour()
nn.train(X_train_rows,y_train)
nn_pred = nn.predict(X_test_rows)

print(f"accuracy : {np.mean(nn_pred==y_test)}")
```
